```python
#Q6.Write a program to demonstrate the use of logical operators
# Taking input from the user
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

# Demonstrating logical AND
print("\nLogical AND operator:")
if a > 0 and b > 0:
    print("Both numbers are positive.")
else:
    print("At least one number is not positive.")

# Demonstrating logical OR
print("\nLogical OR operator:")
if a > 0 or b > 0:
    print("At least one number is positive.")
else:
    print("Both numbers are non-positive.")

# Demonstrating logical NOT
print("\nLogical NOT operator:")
if not (a < 0 or b < 0):
    print("No number is negative.")
else:
    print("At least one number is negative.")
```

```python
#Q14.Write a program to reverse a string using a while loop
# Program to reverse a string using while loop

string = input("Enter a string: ")
reversed_string = ""
i = len(string) - 1          # Start from the last index

while i >= 0:
    reversed_string += string[i]
    i -= 1

print("Reversed string:", reversed_string)
```

```python
#Q15.Write a python program to calculate the factorial of a number provided by the user using a while loop
user_response=int(input("enter the number whose factorial is to be generated"))
factorial=1
i=1
while i <= user_response:
    factorial=factorial*i
    i=i+1
    print(factorial)
```
